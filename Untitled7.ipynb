```python

import torch
from transformers import pipeline

asr = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small",
    device=0,
    torch_dtype=torch.float16
)

print("Whisper-small loaded successfully!")
```

```python

def transcribe_sample(dataset, language):
    audio = dataset[0]["audio"]

    result = asr(
        {
            "array": audio["array"],
            "sampling_rate": audio["sampling_rate"]
        },
        generate_kwargs={
            "language": language,
            "task": "transcribe",
            "num_beams": 1
        }
    )

    print("Language:", language)
    print("Reference:", dataset[0]["transcription"])
    print("Prediction:", result["text"])

transcribe_sample(tamil_ds, "tamil")
transcribe_sample(english_ds, "english")
```

```python

def transcribe_direct(audio_item, language, num_beams=1):
    features = processor(
        audio_item["array"],
        sampling_rate=audio_item["sampling_rate"],
        return_tensors="pt"
    ).input_features.to(
        device=model.device,
        dtype=model.dtype
    )

    with torch.inference_mode():
        token_ids = model.generate(
            input_features=features,
            language=language,       # Pass string, not token ID
            task="transcribe",
            num_beams=num_beams
        )

    return processor.batch_decode(
        token_ids,
        skip_special_tokens=True
    )[0].strip()
```

```python
import time
import pandas as pd

results = []

for i, item in enumerate(examples):
    language_code = (
        "tamil" if item["language"] == "Tamil" else "english"
    )

    audio = item["audio"]
    duration = len(audio["array"]) / audio["sampling_rate"]

    row = {
        "id": item["id"],
        "language": item["language"],
        "reference": item["reference"],
        "audio_duration_sec": duration
    }

    # Compare two decoding methods
    for method, beams in [("baseline", 5), ("greedy", 1)]:
        start = time.perf_counter()

        prediction = transcribe_direct(
            audio,
            language_code,
            num_beams=beams
        )

        elapsed = time.perf_counter() - start

        row[f"{method}_prediction"] = prediction
        row[f"{method}_latency_sec"] = elapsed
        row[f"{method}_rtf"] = elapsed / max(duration, 1e-9)

    results.append(row)

    if (i + 1) % 10 == 0:
        pd.DataFrame(results).to_csv(
            "/content/asr_results.csv", index=False
        )
        print(f"Processed {i + 1}/{len(examples)} examples")

df = pd.DataFrame(results)
df.to_csv("/content/asr_results.csv", index=False)

print("\nEvaluation completed!")
print("Total examples:", len(df))
display(df[[
    "id", "language", "reference",
    "baseline_prediction", "greedy_prediction"
]].head(10))
```

```python
!pip -q install jiwer

from jiwer import wer, cer
import pandas as pd

summary = []

for language in ["Tamil", "English"]:
    subset = df[df["language"] == language]

    for method in ["baseline", "greedy"]:
        summary.append({
            "Language": language,
            "Method": method,
            "WER": wer(
                subset["reference"].tolist(),
                subset[f"{method}_prediction"].tolist()
            ),
            "CER": cer(
                subset["reference"].tolist(),
                subset[f"{method}_prediction"].tolist()
            ),
            "Mean latency (seconds)": subset[
                f"{method}_latency_sec"
            ].mean(),
            "Mean RTF": subset[f"{method}_rtf"].mean()
        })

metrics_df = pd.DataFrame(summary)
display(metrics_df.round(4))

metrics_df.to_csv("/content/asr_metrics.csv", index=False)
print("Metrics saved successfully.")
```

```python
import numpy as np
import pandas as pd
import time
from jiwer import wer, cer

def add_noise_snr(audio_array, snr_db=15, seed=42):
    audio = np.asarray(audio_array, dtype=np.float32)
    rng = np.random.default_rng(seed)

    signal_power = np.mean(audio ** 2)
    if signal_power < 1e-12:
        return audio.copy()

    noise = rng.normal(size=audio.shape).astype(np.float32)
    noise_power = np.mean(noise ** 2)

    target_noise_power = signal_power / (10 ** (snr_db / 10))
    noise = noise * np.sqrt(target_noise_power / noise_power)

    return (audio + noise).astype(np.float32)

noisy_results = []

for i, item in enumerate(examples):
    language_code = (
        "tamil" if item["language"] == "Tamil" else "english"
    )
    audio = item["audio"]

    noisy_audio = add_noise_snr(
        audio["array"], snr_db=15, seed=1000 + i
    )

    start = time.perf_counter()

    noisy_prediction = transcribe_direct(
        {
            "array": noisy_audio,
            "sampling_rate": audio["sampling_rate"]
        },
        language_code,
        num_beams=5
    )

    elapsed = time.perf_counter() - start

    noisy_results.append({
        "id": item["id"],
        "language": item["language"],
        "reference": item["reference"],
        "noisy_prediction": noisy_prediction,
        "noisy_latency_sec": elapsed
    })

    if (i + 1) % 10 == 0:
        print(f"Processed noisy audio: {i + 1}/100")

noisy_df = pd.DataFrame(noisy_results)
noisy_df.to_csv("/content/noisy_asr_results.csv", index=False)

print("Noisy-audio evaluation completed.")
```

```python
comparison = []

for language in ["Tamil", "English"]:
    clean = df[df["language"] == language]
    noisy = noisy_df[noisy_df["language"] == language]

    comparison.append({
        "Language": language,
        "Condition": "Clean",
        "WER": wer(
            clean["reference"].tolist(),
            clean["baseline_prediction"].tolist()
        ),
        "CER": cer(
            clean["reference"].tolist(),
            clean["baseline_prediction"].tolist()
        )
    })

    comparison.append({
        "Language": language,
        "Condition": "Noisy (15 dB SNR)",
        "WER": wer(
            noisy["reference"].tolist(),
            noisy["noisy_prediction"].tolist()
        ),
        "CER": cer(
            noisy["reference"].tolist(),
            noisy["noisy_prediction"].tolist()
        )
    })

comparison_df = pd.DataFrame(comparison)
display(comparison_df.round(4))

comparison_df.to_csv(
    "/content/noise_comparison.csv", index=False
)
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

LLM_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(LLM_NAME)

llm = AutoModelForCausalLM.from_pretrained(
    LLM_NAME,
    torch_dtype=torch.float16
).to("cuda")

llm.eval()

print("LLM loaded successfully!")
```

```python

def generate_response(transcript, language):
    messages = [
        {
            "role": "system",
            "content": (
                f"You are a helpful speech assistant. "
                f"Respond in {language}. Be concise. "
                "Do not invent details unsupported by the input. "
                "If the input is unclear, ask for clarification."
            )
        },
        {
            "role": "user",
            "content": transcript
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(llm.device)

    with torch.inference_mode():
        output = llm.generate(
            **inputs,
            max_new_tokens=100,
            do_sample=False
        )

    new_tokens = output[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()


print("Tamil response:")
print(generate_response(df.iloc[0]["baseline_prediction"], "Tamil"))

print("\nEnglish response:")
print(generate_response(df.iloc[50]["baseline_prediction"], "English"))
```

```python

import time
import pandas as pd

pipeline_results = []

for i, row in df.iterrows():
    language = row["language"]
    transcript = row["baseline_prediction"]

    start = time.perf_counter()
    response = generate_response(transcript, language)
    llm_latency = time.perf_counter() - start

    pipeline_results.append({
        "id": row["id"],
        "language": language,
        "reference_transcript": row["reference"],
        "asr_transcript": transcript,
        "llm_response": response,
        "asr_latency_sec": row["baseline_latency_sec"],
        "llm_latency_sec": llm_latency,
        "end_to_end_latency_sec": (
            row["baseline_latency_sec"] + llm_latency
        ),
        "audio_duration_sec": row["audio_duration_sec"]
    })

    if (i + 1) % 10 == 0:
        pd.DataFrame(pipeline_results).to_csv(
            "/content/pipeline_results.csv", index=False
        )
        print(f"Processed {i + 1}/100")

pipeline_df = pd.DataFrame(pipeline_results)
pipeline_df.to_csv("/content/pipeline_results.csv", index=False)

print("Pipeline evaluation completed!")
```

```python
latency_summary = pipeline_df.groupby("language").agg(
    samples=("id", "count"),
    mean_asr_latency_sec=("asr_latency_sec", "mean"),
    mean_llm_latency_sec=("llm_latency_sec", "mean"),
    mean_end_to_end_latency_sec=(
        "end_to_end_latency_sec", "mean"
    )
).reset_index()

display(latency_summary.round(3))

latency_summary.to_csv(
    "/content/pipeline_latency_summary.csv",
    index=False
)

print("Latency summary saved.")
```
